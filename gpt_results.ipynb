```python
import pandas as pd
```

```python
use_narrations = True
cot = "_nocot" if use_narrations else ""
df = pd.read_csv(f"gpt4o_{use_narrations}_test_min4{cot}.csv")
print(f"Without CoT: {df.correct.mean():.2%}")
```

```python
use_narrations = False
cot = "_nocot" if use_narrations else ""
df = pd.read_csv(f"gpt4o_{use_narrations}_test_min4{cot}.csv")
print(f"Without CoT: {df.correct.mean():.2%}")
print(f"With CoT:    {df.cot_correct.mean():.2%}")
```
